### Opret chatbot ###

```python
client = openai.OpenAI()
```

###### max tokens = 250 (hvor mange tokens der må bruges i svaret)
###### temperature til 0 (mest kreativitet)
###### seed til 365 (for at få samme svar hver gang)

```python
completion = client.chat.completions.create(
    model="gpt-4",
    messages=[{"role": "system",
               "content": "You are Marv, a chatbot that, reluctantly answers questions            with sarcastic responses."},
              {"role": "user",
               "content": "Explain what a black hole is?"}],
    max_tokens=250,
    temperature=0.0,
    seed=365,
    stream=True)
```

```python
print(completion.choices[0].message.content)
```

```python
print(completion.choices[0].message.content)
```

###### Iterer over Stream svar fra ChatBot ######

```python
for chunk in completion:
    if chunk.choices[0].delta.content is not None:
        print(chunk.choices[0].delta.content, end="")
```

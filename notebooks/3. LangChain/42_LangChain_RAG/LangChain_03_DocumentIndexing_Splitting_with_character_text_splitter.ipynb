###### Import ######

```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_text_splitters import CharacterTextSplitter
import copy
```

##### doc2txt instans med doc filnavn som argument #####

```python
loader = Docx2txtLoader("Introduction_to_Data_and_Data_Science.docx")
```

###### Fjern newline characters fra tekst ######

```python
# Loader dokument
pages = loader.load()
# Rens tekst ved at fjerne newline characters
for i in range(len(pages)):
    pages[i].page_content = ' '.join(pages[i].page_content.split())
```

```python
# Antal af characters i dokument
len(pages[0].page_content)
```

###### Split dokument i mindre chunks med Character Text Splitter ######

```python
# Opret Character Text Splitter instans
char_splitter = CharacterTextSplitter(separator="",
                                      chunk_size=500,
                                      chunk_overlap=0)
```

```python
# Split dokument i mindre chunks
pages_char_split = char_splitter.split_documents(pages)
```

### Indexing: Inspecting and Managing Documents in a Vectorstore ###

###### Setup API Key som miljøvariabel ######

```python
%load_ext dotenv
%dotenv ../.env
```

###### Import ######

```python
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

###### Opret Embedding Model instans ######

```python
embedding = OpenAIEmbeddings(model = 'text-embedding-ada-002')
```

###### Opret Chroma Vectorstore ######

```python
vectorstore_from_directory = (Chroma(persist_directory="./chroma_db",
                                     embedding_function=embedding))
```

###### GetAll - Hente ALLE embeddings fra fra db - Get() ######

```python
vectorstore_from_directory.get()
```

###### Get - Hent specifik embedding - Get(ids=...) ######

```python
vectorstore_from_directory.get(ids="8210f730-93a1-434f-8385-360fa1ab7bf2",
                               include=["embeddings"])
```

###### Create - Tilføj dokument til db ######

```python
document_to_add = Document(page_content="Dette er en Test",
                           metadata={'Course Title': 'Introduction to Data and Data Science',
                                     'Lecture Title': 'Test Lecture'})
```

```python
vectorstore_from_directory.add_documents([document_to_add])
```

```python
vectorstore_from_directory.get(ids='ecc17bc4-5ee7-417f-a09e-26ba16a9a364')
```

###### Update - Opdatere et dokument i db ######

```python
document_to_update = Document(page_content="Dette er en OPDATERET Test",
                           metadata={'Course Title': 'Introduction to Data and Data Science',
                                     'Lecture Title': 'Updated Test Lecture'})
```

```python
vectorstore_from_directory.update_document(document_id="ecc17bc4-5ee7-417f-a09e-26ba16a9a364",
                                           document = document_to_update)
```

```python
vectorstore_from_directory.get(ids='ecc17bc4-5ee7-417f-a09e-26ba16a9a364')
```

###### Delete - Slette et dokument fra db ######

```python
vectorstore_from_directory.delete("ecc17bc4-5ee7-417f-a09e-26ba16a9a364")
```

###### Import ######

```python
from langchain_openai import ChatOpenAI

from langchain_core.messages import HumanMessage, AIMessage
```

###### Opret ChatOpenAI  med gpt-4, model_kwargs ######

```python
chat = ChatOpenAI(
    model_name='gpt-4',
    seed=365,
    temperature=0,
    max_tokens=100
)
```

###### Opret system & Human beskeder og kald chat.invoke() ######

```python
message_h_dog = HumanMessage(content = '''I've recently adopted a dog.
                                    Can you suggest some dog names?''')

message_ai_dog = AIMessage(content = '''Oh, absolutely. Because nothing screams "I'm a responsible pet owner" like asking a chatbot to name your new best friend. How about "Bark Twain" if it's a literary hound, or "Sir Wag-a-lot" for the knightly type? Maybe "Bark Zuckerberg" if it's into social media, or "Biscuit" if it's as bland as your creativity in naming.''')

message_h_cat = HumanMessage(content = '''I've recently adopted a cat.
                                    Can you suggest some cat names?''')

message_ai_cat = AIMessage(content = '''Oh, absolutely. Because nothing screams "I'm a responsible pet owner" like asking a chatbot to name your new best friend. How about "Furry McFurFace", "Sir Meowsalot" or "Catastrophe"?.''')

message_h_fish = HumanMessage(content = '''I've recently adopted a fish.
                                    Can you suggest some fish names?''')
```

###### Invoke med beskeder ######

```python
response = chat.invoke([message_h_dog, message_ai_dog, message_h_cat, message_ai_cat, message_h_fish])
```

###### Print response direkte ######

```python
print(response.content)
```

###### Import ######

```python
from langchain_openai import ChatOpenAI

from langchain_core.messages import HumanMessage, SystemMessage
```

###### Opret ChatOpenAI  med gpt-4, model_kwargs ######

```python
chat = ChatOpenAI(
    model_name='gpt-4',
    seed=365,
    temperature=0,
    max_tokens=100
)
```

```python
message_s = SystemMessage(content = '''You are Marv, a chatbot that reclutlantly                                        answers questions with sarcastic responses.''')

message_h = HumanMessage(content = '''I've recently adopted a dog.
                                    Can you suggest some dog names?''')
```

```python
response = chat.invoke([message_s, message_h])
```

```python
print(response.content)
```
